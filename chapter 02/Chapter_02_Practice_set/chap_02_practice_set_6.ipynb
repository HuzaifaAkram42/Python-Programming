### Q1
Find the length of a given string without using the `len()` function.

```python
s = input("Enter a string ")
counter = 0
for i in s:
    counter += 1
print(counter)
```

### Q2
Find the count of digit of a given string without using the `len()` function.

```python
s = input("Enter a string ")
digit = "0123456789"
counter = 0
for i in s:
    if i in digit:
        counter += 1
print("Digits Count",counter)
```

### Q6
Write a program to check whether a given string is a palindrome or not.

```python
s = input("Enter a string ")
flag = True
for i in range(1,len(s)//2):
    if s[i] != s[len(s)-i-1]:
        print("Not Palindrome")
        flag = False
        break
if flag:
    print("Its a Palindrome")
```

```python
s = input("Enter a string ")
left = 0
right = len(s) - 1
while left < right:
    if s[left] != s[right]:
        print("not a palindrom")
        break
    else:
        left += 1
        right -= 1
else:
    print("Its Palindrome")
```

### Q8
Convert an integer to a string without using `str()`.

```python
number = int(input("Enter a number "))
digit = "0123456789"
result = ''
while number != 0:
    result = digit[number % 10] + result
    number = number // 10
result
```

### Q10
Convert a string to title case (first letter of each word capitalized, rest lowercase) without using `title()`.

```python
s = input("Enter a string ")
L = []
for i in s.split():
    L.append(i[0].upper() + i[1:].lower())
print(" ".join(L))
```

### Q11
Reverse a given string without using slicing (`[::-1]`) or the `reversed()` function.

```python
s = input("Enter a String ")
result = ""
for i in s:
    result = i + result
print("Original Sentence:",s)
print("reversed Sentence:",result)
```

### Q12
Count the number of vowels and consonants in a given string.

```python
s = input("Enter a string ")
vowel = ['a','e','i','o','u']
counter_v = 0
counter_c = 0
counter_d = 0
digit = '0123456789'
for i in s:
    if i != " ":
        if i.lower() in vowel:
            counter_v += 1
        elif i in digit:
            counter_d += 1
        else:
            counter_c += 1
    else:
        pass
print("Consonants Counts",counter_c)
print("Vowel Counts",counter_v)
```

### Q13
Count the number of uppercase and lowercase letters in a given string, without using `isupper()` or `islower()`.

```python
s = input("Enter a string ")

upper_count = 0
lower_count = 0

for char in s:
    if "A" <= char <= "Z":
        upper_count += 1
    elif "a" <= char <= "z":
        lower_count += 1
print("Uppercase letters:", upper_count)
print("Lowercase letters:", lower_count)
```

### Q14
Remove duplicate characters from a string while preserving the original order of first appearance.

```python
s = input("Enter a string ")
result = ''
for char in s:
    if char not in result:
        result = result + char
print(result)
        
```
